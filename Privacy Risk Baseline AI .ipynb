```python
print("========== DATA QUALITY CHECK ==========\n")

print("Missing values in each column:")
display(df.isnull().sum())
```

```python
print("========== BASIC DATA ANALYSIS ==========\n")

print("Average Age:", df["Age"].mean())
print("Average Income:", df["Income"].mean())
print("Average Attendance:", df["Attendance"].mean())

print("\nMedical Condition Distribution:")
display(df["Medical_Condition"].value_counts())

print("\nResult Distribution:")
display(df["Result"].value_counts())
```

```python
print("========== AI MODEL TRAINING ==========\n")

# Select features
X = df[["Age", "Income", "Attendance"]]

# Convert Pass/Fail to numbers
y = df["Result"].map({
    "Pass": 1,
    "Fail": 0
})

# Split dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)

# Create Decision Tree
model = DecisionTreeClassifier(
    max_depth=3,
    random_state=42
)

# Train model
model.fit(X_train, y_train)

# Predict
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Model: Decision Tree")
print("Features used:", list(X.columns))
print("Model Accuracy:", round(accuracy * 100, 2), "%")
```

```python
print("========== PRIVACY LEAKAGE PATHS ==========\n")

leakage_paths = [
    "1. Dataset files can expose Name, Email and Student ID.",
    "2. Logs may accidentally print personal information.",
    "3. Model outputs may reveal sensitive information.",
    "4. Screenshots may expose identifiable information.",
    "5. Unauthorized users may access stored dataset files.",
    "6. Age + City + Income can potentially help re-identify a person.",
    "7. Medical information can be exposed through analysis."
]

for item in leakage_paths:
    print(item)
```

```python
print("========== PRIVACY RISK GRAPH ==========")

risk_count = risk_table["Risk_Level"].value_counts()

plt.figure(figsize=(7, 5))

risk_count.plot(kind="bar")

plt.title("Baseline Privacy Risk Distribution")
plt.xlabel("Risk Level")
plt.ylabel("Number of Fields")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```
