```python
!pip -q install dnspython

import dns.resolver
import pandas as pd

# 5 websites and DNS record types
queries = [
    ("youtube.com", "A"),
    ("linkedin.com", "AAAA"),
    ("microsoft.com", "A"),
    ("github.com", "MX"),
    ("amazon.com", "A")
]

results = []

for domain, record_type in queries:
    try:
        answers = dns.resolver.resolve(domain, record_type)
        ttl = answers.rrset.ttl

        for answer in answers:
            results.append({
                "Domain Name": domain,
                "Query Type": record_type,
                "Response": str(answer),
                "TTL": ttl
            })

    except Exception as e:
        results.append({
            "Domain Name": domain,
            "Query Type": record_type,
            "Response": "No Response",
            "TTL": "-"
        })

df = pd.DataFrame(results)

print("DNS TRAFFIC INSPECTION")
print("=" * 80)

display(df)
```
